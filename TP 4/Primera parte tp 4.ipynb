```python
# Unión de las dos bases
df_unido = pd.concat([df_2004_noa, df_2024_noa], ignore_index=True)
```

```python
# definicion de la base de datos "respondieron" para tener los datos ordenados desde el principio.
respondieron = df_unido[(df_unido["ESTADO"].notna()) & (df_unido["ESTADO"] != 0)]
```

```python
from sklearn.model_selection import train_test_split
import statsmodels.api as sm
```

```python
# Copia para trabajar con seguridad
respondieron = respondieron.copy()

# Crear variable dependiente: 1 si desocupado, 0 si no
respondieron["desocupado"] = (respondieron["ESTADO"] == 2).astype(int)

# Variables de interés seleccionadas
variables_interes = [
    "CH04",  # Sexo
    "CH06",  # Edad
    "ESTADO",  # Condición de actividad
    "CAT_OCUP", # Categoría ocupacional
    "AGLOMERADO", # Código de Aglomerado
    "NIVEL_ED",  # Nivel educativo
    "P47T",  # Ingreso total individual
    "P21",  # Ingreso ocupación principal
    "CH03",  # Parentesco
    "CH07",  # Estado civil
    "CH08",  # Cobertura médica
    "CH10",  # Asistencia educativa
    "CH11",  # Tipo de establecimiento educativo
    "CH12",  # Nivel cursado
    "CH13",  # Nivel finalizado
]

# Subconjunto X e y
X = respondieron[variables_interes]
y = respondieron["desocupado"]

# División train/test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=444)

# Conversión segura a numérico
X_train = X_train.apply(pd.to_numeric, errors='coerce')
X_test = X_test.apply(pd.to_numeric, errors='coerce')

# Imputar valores faltantes con 0
X_train = X_train.fillna(0)
X_test = X_test.fillna(0)

# Agregar columna de unos
X_train = sm.add_constant(X_train, has_constant='add')
X_test = sm.add_constant(X_test, has_constant='add')

# Tabla de diferencias de medias
mean_train = X_train.mean()
mean_test = X_test.mean()
diff_means = pd.DataFrame({
    "Media Train": mean_train,
    "Media Test": mean_test,
    "Diferencia": mean_train - mean_test
})

# Mostrar tabla
print(diff_means)
```

La tabla presenta las medias de las variables explicativas seleccionadas en los conjuntos de entrenamiento (train) y prueba (test), así como la diferencia entre ellas. Las diferencias que se observan son, en general, bastante pequeñas, lo que sugiere que la partición aleatoria realizada con train_test_split (usando random_state=444) fue adecuado.

Algunos puntos a resaltar:

Las variables como CH04 (sexo), CH06 (edad), CH07 (estado civil) y otras variables categóricas como CH08 (cobertura médica) y CH10 (asistencia educativa) muestran diferencias menores a 0.03 puntos, lo que indica que ambas muestras representan bien al total de la población.

En cuanto a las variables numéricas más sensibles, tenemos:

P47T (ingreso total individual): diferencia ≈ 2.788 pesos

P21 (ingreso de la ocupación principal): diferencia ≈ 3.370 pesos

Aunque estas diferencias pueden parecer un poco más grandes en términos absolutos, hay que ponerlas en perspectiva, ya que los montos promedios rondan entre 75.000 y 78.000 pesos. Por lo tanto, en términos relativos, la diferencia es inferior al 5%, lo cual sigue siendo aceptable para un muestreo aleatorio.

AGLOMERADO muestra una diferencia promedio de ≈ 0.38. Dado que esta variable es un identificador geográfico (y no ordinal), esta diferencia no tiene una interpretación directa en cuanto a magnitud o sesgo.

En resumen, no se han encontrado indicios de que el conjunto de entrenamiento esté sesgado en comparación con el conjunto de prueba, lo que nos permite avanzar con el entrenamiento del modelo con confianza en la representatividad de los datos.
